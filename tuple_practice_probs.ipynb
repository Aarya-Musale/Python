```python
#17. Write a Python program to apply a square function to every item in a tuple using map(), and also demonstrate the equivalent generator expression approach.


#Method1
numbers = (1, 2, 3, 4, 5, 6)
a = tuple(map(lambda x:x*x,numbers))
print('Squared:',a)



#Method2
numbers = (1, 2, 3, 4, 5, 6)
b = tuple(x*x for x in numbers)
print('Squared:',b)
```

```python
#19. Write a Python program to find all elements that are common to two different tuples.

t1 = (1, 2, 3, 4, 5, 6)
t2 = (4, 5, 6, 7, 8, 9)
a = set(t1)
b = set(t2)

print('Common Elements:',tuple(a.intersection(b)))
```

```python
#20. Write a Python program to “modify” a tuple by converting it to a list, changing a specific item, and converting it back to a tuple.

colours = ("red", "green", "blue")  #replace "green" with "yellow"
print('Original:',colours)

a = list(colours)
a[1] = 'yellow'
print('Modified:',tuple(a))
```
