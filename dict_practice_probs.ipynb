```python
#19. Write a Python program to create a new dictionary containing only the key-value pairs from an existing dictionary where the value meets a specified condition.

scores = {"Alice": 82, "Bob": 45, "Carol": 91, "Dave": 58, "Eve": 73}
s = {k:v for k,v in scores.items() if v>60}
s
```

```python
#20. Write a Python program to find the key associated with the smallest numerical value in a dictionary.

stock = {"apples": 34, "bananas": 12, "oranges": 57, "grapes": 8, "mangoes": 23}
s = min(stock,key=stock.get)
print('Lowest Stock Item:',s)
```

```python
#21. Write a Python program to find the key associated with the highest numerical value in a dictionary.

scores = {"Alice": 88, "Bob": 95, "Carol": 72, "Dave": 95, "Eve": 84}
s = max(scores,key=scores.get)
print('Top Scorer:',s)
```

```python
#26. Write a Python program to count the frequency of each word in a string, treating words case-insensitively so that “The” and “the” are counted as the same word.

text = "the cat sat on the mat the cat"

import collections
text_lower = text.lower()
text_split = text.split()
c = collections.Counter(text_split)
c
```
