```python
#1. Find the Largest Element

nums = [1,2,10,52,63,21,45,59,87,85,35]

largest = nums[0]

for i in nums:
  if i > largest:
    largest = i

print('The Largest number is:',largest)
```

```python
#2. Find the Smallest Element

nums = [4,2,6,8,10,52,64,14]

smallest = nums[0]

for i in nums:
  if i < smallest:
    smallest = i

print('The Smallest number is:',smallest)
```

```python
#62. Write a function that removes all negative numbers from a list without creating a new list. You must modify the original list object directly

def remove_neg(lst):
  new_list = []
  for i in lst:
    if i >= 0:
      new_list.append(i)
  return new_list


lst1 = [10, -5, 3, -1, 0, -8, 7]
result = remove_neg(lst1)
print('Modified List:',result)
```
